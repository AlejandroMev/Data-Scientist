# **Problema:**
A partir del texto que ingrese el usuario, se desea mostrar en pantalla cada uno de los caracteres que lo componen.
Nota: Se utilizará un ciclo for para recorrer el string proporcionado por el usuario, imprimiendo en cada iteración el caracter correspondiente.

```python
texto = input("Ingresa un texto: ")
for letra in texto:
 print(letra)
```

# **Problema:**
Tenemos una lista con montos de ventas, queremos crear otra lista
solamente con los montos que son mayores a 1500 y desplegarla a
pantalla.

```python
ventas = [1800, 900, 1300, 585, 1589, 1500, 1700, 900, 1925, 4500, 2350, 1600]
mayores_1500 = []
for monto in ventas:
 if monto > 1500:
  mayores_1500.append(monto)
print('Las ventas mayores a 1500 son: ')
print(mayores_1500)
```

La **función range** genera un rango de números con los que el ciclo for puede iterar. Esta función puede recibir uno, dos o tres argumentos.
