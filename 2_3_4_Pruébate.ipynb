```python
# 1. Lectura de archivo

import pandas as pd
archivo1 = pd.read_csv("/content/4-vaccination_tweets.csv")
archivo1.head(8)
```

```python
# 2. Estructura del archivo

archivo1.dtypes
```

```python
# 3. Selección de diferentes columnas

archivoColumnas = archivo1[["user_name", "user_description", "user_followers", "source"]]
archivoColumnas
```

```python
# 4. Filtrado por condición

filtroarchivo = archivo1[archivo1["user_followers"] > 5000]
filtroarchivo
```

```python
# 5. Filtro de columnas con condición

Filtro2 = archivo1[archivo1["user_friends"] > 100][["user_name", "user_location", "user_verified"]]
Filtro2
```

```python
# 6. Selección de únicamente datos numéricos

datonumerico = archivo1.select_dtypes(include = "number")
datonumerico
```

```python
# 7. Condicionales

Condicion1 = archivo1[(archivo1["source"] == "Twitter for iPhone") |
                      (archivo1["source"] =="Twitter for Android")]
Condicion1
```

```python
# 8. Condicionales con where

numericos = archivo1.select_dtypes(include=['number'])
numericos.where(numericos>1000, 0)
```

```python
# 9. Uso de comando query

archivo1.query('user_followers>user_friends & user_favourites>user_friends')
```

```python
# 10. elección de una muestra aleatoria

archivo1.sample(frac=0.25)
```
