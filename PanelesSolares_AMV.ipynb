1. Solicita al usuario la medida de los dos lados que delimitan la superficie disponible para los paneles en metros.

```python
Lado_A = float(input("Bienvenido(a), por favor ingrese el lado A de la superficie en metros: "))
Lado_B = float(input("Por favor ingrese el lado B de la superficie en metros: "))
```

2. Solicita al usuario el consumo promedio de energía mensual del hogar en kWh.

```python
consumo = input("Por favor ingrese el consumo promedio de energía mensual del hogar en kWh: ")
```

3. Pide al usuario el porcentaje de eficiencia de los paneles solares que quiere comprar. Debe ser un valor entre 0 y 1. Por ejemplo: 0.5 corresponde al 50%.

```python
eficiencia = float(input("Por favor ingrese el porcentaje de eficiencia en decimales de los paneles solares que quiere comprar (entre 0.0 y 1.0): "))
print("Su eficiencia seleccionada es del", (eficiencia*100),"%")
```

4. Toma en cuenta que México cuenta una irradiación solar media de 6.36 kWh/m2 por día (ASOLMEX, 2021)

```python
irradicion_media = 6.36
```

5. Calcula el área de la superficie disponible para instalar los paneles solares. Considera la fórmula: area = lado1 * lado2

```python
area_superficie = float(Lado_A) * float(Lado_B)
print("El área de la superficie disponible para instalar los paneles solares es de", area_superficie, "metros cuadrados")
```

6. Calcula la producción energética por mes si se instalan los paneles en la superficie disponible. Piensa que el mes tiene 30 días  y la fórmula: produccion_mes = area * irradiacion * dias_mes * porcentaje_eficiencia.

```python
produccion_mes = float(area_superficie) * float(irradicion_media) * 30 * float(eficiencia)
print("La producción energética si se instalan los paneles en la superficie disponible es de", produccion_mes, "kWh mensuales")
```

7. Determina la viabilidad de instalar los paneles, considerando que para que sea conveniente instalarlos la producción energética por mes debe ser igual o mayor 120% del consumo mensual del cliente.

```python
viabilidad = float(produccion_mes) >= 1.2 * float(consumo)
print("La viabilidad de instalar los paneles es:", viabilidad)
```

8. Imprime un reporte de los resultados que incluya: el área disponible para instalar los paneles solares, el consumo promedio mensual de energía del hogar, el porcentaje de eficiencia de los paneles, la constante de irradiación media considerada, la producción energética estimada por mes y la viabilidad de la instalación. Asegúrate que el despliegue de los datos numéricos sea con máximo dos cifras decimales.

```python
print("Reporte de resultados:")
print(f"Área disponible para instalar los paneles: {area_superficie:.2f} metros cuadrados")
print(f"Consumo promedio mensual de energía del hogar: {float(consumo):.2f} kWh")
print(f"Porcentaje de eficiencia de los paneles: {eficiencia:.2f}")
print(f"Constante de irradiación media considerada: {irradicion_media:.2f} kWh/m2 por día")
print(f"Producción energética estimada por mes: {produccion_mes:.2f} kWh mensuales")
print("Viabilidad de la instalación:", viabilidad)
```
